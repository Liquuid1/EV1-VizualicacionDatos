# 1) Entender el negocio

## Contexto del caso

**StreamView Analytics** es una empresa de streaming que distribuye películas y series en distintos mercados de Latinoamérica. Durante los últimos años, la empresa ha ampliado considerablemente su catálogo y dispone de información sobre títulos, directores, actores, países, idiomas, géneros, clasificación, popularidad, valoración, presupuesto e ingresos.

Actualmente, la información se encuentra en reportes independientes construidos con criterios diferentes. Esto dificulta obtener una visión consolidada del catálogo, comparar resultados entre áreas y transformar los datos en decisiones estratégicas.

## Problema de negocio

La empresa necesita convertir los datos de su catálogo audiovisual en información clara, consistente y accionable para apoyar las decisiones de adquisición, gestión de contenidos, marketing y planificación estratégica.

Los principales problemas identificados son:

- No existe una visión consolidada del catálogo disponible.
- Es difícil identificar rápidamente los géneros con mayor popularidad.
- Las decisiones de adquisición se realizan utilizando información parcial.
- No existe una forma simple de comparar países productores.
- Marketing no dispone de información consolidada para priorizar contenidos y segmentos en sus campañas.
- El área de Adquisición y Gestión de Contenidos necesita conocer las características de los contenidos mejor evaluados y más populares.
- Los reportes existentes presentan exceso de información, problemas de diseño y poca interacción.

## Objetivo del análisis

Diseñar una solución de Visual Analytics que permita explorar el catálogo de StreamView Analytics mediante visualizaciones claras y dashboards interactivos, entregando evidencia para apoyar decisiones de adquisición, promoción y posicionamiento de contenidos.

## Público objetivo y necesidades

El público principal de este análisis es la **Gerencia de Contenidos y Adquisición**, que necesita decidir qué contenidos adquirir, renovar, producir o posicionar.

También utilizarán los resultados:

- **Directorio y Gerencia General:** requieren indicadores consolidados, tendencias, comparaciones y recomendaciones ejecutivas.
- **Gerencia de Marketing:** necesita identificar contenidos populares, categorías con potencial y títulos prioritarios para campañas.
- **Gerencia de Producto:** busca comprender el catálogo para mejorar la experiencia de navegación de los usuarios.
- **Equipo Data & Analytics:** debe mantener dashboards confiables y asegurar la calidad de la información.

## Datos disponibles

El análisis se realizará exclusivamente con los datasets entregados por StreamView Analytics:

1. **Películas:** contiene información de identificación, título, director, reparto, país, fecha de incorporación, año de estreno, clasificación, duración, géneros, idioma, popularidad, votos, calificación, presupuesto e ingresos.
2. **Series:** contiene información de identificación, título, director, reparto, país, fecha de incorporación, año de estreno, duración, géneros, idioma, popularidad, votos y calificación.

Las variables `budget` e `revenue` solo están disponibles para películas, por lo que los indicadores financieros deben analizarse separadamente y no deben utilizarse para comparar directamente películas con series.

## Preguntas clave del negocio

- ¿Cómo se distribuye el catálogo entre películas y series?
- ¿Qué géneros, países e idiomas concentran la mayor cantidad de contenidos?
- ¿Cómo ha evolucionado el catálogo y la incorporación de títulos durante los últimos años?
- ¿Qué contenidos presentan mayor popularidad y cuáles tienen mejores calificaciones?
- ¿Qué características comparten los contenidos más populares y mejor evaluados?
- ¿Existe relación entre popularidad, calificación y cantidad de votos?
- ¿Qué países representan oportunidades para futuras adquisiciones?
- ¿Qué películas obtienen mayores ingresos o retorno sobre el presupuesto?
- ¿Qué contenidos y categorías deberían priorizarse en campañas promocionales?

## Indicadores iniciales (KPIs)

- Total de contenidos, películas y series.
- Cantidad de países, idiomas y géneros representados.
- Popularidad promedio y contenidos más populares.
- Calificación promedio y contenidos mejor evaluados.
- Distribución del catálogo por género, país, idioma y clasificación.
- Contenidos incorporados y estrenos por año.
- Para películas: presupuesto total, ingresos totales y ROI aproximado (`revenue / budget`).

## Decisiones que debe apoyar el análisis

- Priorizar géneros y mercados para futuras adquisiciones.
- Seleccionar contenidos para campañas de marketing.
- Identificar oportunidades de posicionamiento del catálogo.
- Orientar decisiones de renovación, producción y gestión de contenidos.
- Comunicar hallazgos de forma clara a las distintas áreas de la organización.

> **Alcance y limitaciones:** la popularidad es un índice relativo y no representa reproducciones. Los indicadores financieros deben considerar únicamente registros con presupuesto e ingresos disponibles y válidos.
